```python
import numpy as np
import pandas as pd
from statistics import mode
arr = np.array([10, 20, 30, 40, 50])
print("Array:",arr)
mean=np.mean(arr)
median=np.median(arr)
print("Mean:",mean)
print("Median:",median)
print("Mode:",mode(arr))
print("Reshaped 2D:\n",arr.reshape(5,1))
print("Sliced (index 1 to 3):", arr[1:4])
#---Create a DataFrame--
data={"Name":["Asha","Ravi","Meera","Karan"],
      "Marks":[98,97,92,89],
      "Branch":["CSE","CSE","ECE","CSE"]}
df=pd.DataFrame(data)
print(df)
#---Indexing and slicing---
print("\nFirst 2 rows:\n",  df.iloc[0:2])
print("\nMarks column:\n", df["Marks"])
print("\nRows where Marks > 90:\n", df[df["Marks"]>90])
```
