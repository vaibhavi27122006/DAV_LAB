```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
pd.set_option('future.no_silent_downcasting', True)
df=pd.read_csv("/content/student_performance_dataset.csv")

print("First 5 rows of dataset:")
print(df.head())
print(r"\Dataset Info:")
print(df.info())
print("\n Missing values before handling:")
print(df.isnull().sum())
df.loc[0:5, 'Study_Hours_per_Week']=None
df.loc[0:5, 'Attendence_Rate']=None
df.loc[0:5, 'Past_Exam_Scores']=None
df.loc[0:5, 'Final_Exam_Score']=None
print("\n Missing values after inserting sample NaNs:")
print(df.isnull().sum()[df.isnull().sum() > 0])
df['Study_Hours_per_Week']=df['Study_Hours_per_Week'].fillna(df['Study_Hours_per_Week'].mean())
df['Attendance_Rate']=df['Attendence_Rate'].fillna(df['Attendance_Rate'].median())
df['Past_Exam_Score']=df['Past_Exam_Scores'].ffill()
df['Final_Exam_Score']=df['Final_Exam_Score'].fillna(df['Final_Exam_Score'])
print("\nMissing values after imputation:")
print(df.isnull().sum())
df=pd.get_dummies(df,columns=['Parental_Education_Level'])
print("\nOne-Hot Encoded columns:")
print(df.columns.tolist())
le=LabelEncoder()
df['Pass_Fail_Encoder']=le.fit_transform(df['Pass_Fail'])
print("\nLabel Encoder Result:")
print(df[['Pass_Fail','Pass_Fail_Encoder']].head())
df['Gender_Encoder']=le.fit_transform(df['Gender'])
df['Internet_Access_Encoder']=le.fit_transform(df['Internet_Access_at_Home'])
df['Extracurricular_Encoder']=le.fit_transform(df['Extracurricular_Activities'])
df.to_csv("student_performance_cleaned.csv",index=False)
print("\nCleaned dateset saved as 'student_performance_cleaned.csv'")




```
