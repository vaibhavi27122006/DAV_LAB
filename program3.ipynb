```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("/content/student_data (4).csv")
print("dataset shape:",df.shape)
```

```python
print(df.head())
```

```python
print(df.describe())
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df["G3"], bins=20, kde=True, color="steelblue")
plt.title("Distribution of final grades (G3)")
plt.xlabel("final grades")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="studytime",y="G3",data=df,palette="Set2")
plt.title("Final grade across study time level")
plt.xlabel("study time(1=low, 4=high)")
plt.ylabel("Final grades")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x="G1",y="G3",hue="sex",data=df,palette="coolwarm")
plt.title("Final Period Grade vs Final grade")
plt.xlabel("G1 (Period 1 Grade)")
plt.ylabel(" G3 Final grades")
plt.show()
```

```python
sns.pairplot((df[["age","studytime","G1","G2","G3","sex"]]),hue="sex",palette="husl")
plt.suptitle("pairwise relationships",y=1.02)
plt.show()
```

```python
numeric_cls = ["age", "studytime", "failures", "absences", "G1", "G2", "G3"]
plt.figure(figsize=(8, 6))
sns.heatmap(
    df[numeric_cls].corr(), annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5
)
plt.title("Correlation Heatmap")
plt.show()
```
